**Als je een waxinekaarsje aan steekt en er een glas overheen zet, zal het kaarsje uit gaan. Als je dat waxinekaarsje laat drijven op water terwijl je er het glas er overheen zet, zal het water in het glas stijgen. Op het internet zwerven er verschillende verklaringen voor dit fenomeen rond.**

**1.Het kaarsje verbruikt zuurstof, waardoor er een vacuüm ontstaat en het water omhoog wordt gezogen.**

**2.Het kaarsje verwarmt de lucht in het glas, waardoor de lucht uitzet. Wanneer de kaars dooft koelt de lucht af waardoor de druk afneemt en er water omhoog wordt gezogen.**

**3.Bij het verbranden van de kaars ontstaan er waterdamp en koolstofdioxide. De waterdamp condenseert aan de binnenkant van het glas, waardoor er minder gas in het glas is en de druk afneemt. Hierdoor wordt het water omhoog gezogen.**

## Onderzoek welke verklaring het meest waarschijnlijk is en leg uit waarom de andere verklaringen minder waarschijnlijk zijn.

Wat te verklaren:
Een waxinekaarsje brandt onder een omgekeerd glas dat in water staat.
Het kaarsje dooft en daarna stijgt het waterniveau in het glas.

Theorie:
Druk in het glas wordt bepaald door: pV = nRT

Dit is omdat V in het glas constant blijft en n(hoeveelheid gas) grotendeeks! constant (niet gesloten)
Daarom geldt: p => T (als T daalt, daalt p)

Observaties en resultaat: 
1. Observaties/hypotheses tijdens het branden van kaars:

- De vlam verwarmt lucht in glas -> Temparatuur T van de lucht stijgt
- Bij constant druk zet de lucht uit 
- Een deel van de warme lucht ontsnapt langs de rand van het glas 

Het aantal mol gas n wordt iets kleiner 

2. Observaties na de kaars dooft: 

- Warmtebron verdwijnt 
- Lucht in glas koelt af -> Temparatuur T daalt
- V constant, dus druk p daalt

ideale gaswet: p_eind/p_begin = (n_eind/n_begin) * (T_eind/T_begin) met n_eind < n_begin

Dus druk in glas wordt lager dan de buitenlucht 

3. Eindresultaat:

- Buitenluchtdruk is groter dan druk in glas 
- Water wordt door drukverschil het glas omhoog gedrukt 
- Water stijgt tot p_binnen + pgh = p_buiten, met pgh (hydrostatische druk)

Waarom andere twee verklaringen niet de hoofdreden zijn voor waterstijging:

1. Zuurstofverbruik:

- Lucht bestaat uit 21% zuurstof 
- Een kaars dooft al bij 16% zuurstof 
- Dus slechts enkele procenten van het gas verdwijnt 

Dat levert een te kleine drukdaling om de geobserveerde waterstijging te verklaren 

2. Condenstatie van waterdamp:

- Bij verbranding ontstaat waterdamp dat deels tegen glas condenseert 
- Hierdoor neemt n iets af

Dit draagt ook bij aan drukdaling maar de hoeveelheid n dat afneemt is te klein om de totalte waterstijging te verklaren

Conclusie:
De dominante oorzaak van de waterstijging is de afkoeling van de lucht in het glas na het doven van de kaars
Wat leidt tot een drukdaling volgens de ideale gaswet.
