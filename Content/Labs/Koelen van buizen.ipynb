## Theorie

Volgens Newton's wet van afkoeling is de snelheid waarmee een voorwerp afkoelt evenredig
met het verschil in de temperatuur van het voorwerp ($T$) en de omgeving ($T_0$).
We kunnen dit schrijven als:

$$
\dot{Q} = -hA \left( T(t) - T_0 \right)
$$

waarin:

- $\dot{Q}$ de warmtestroom in $\mathrm{W}$
- $A$ het oppervlak waardoor koeling optreedt in $\mathrm{m^2}$
- $h$ de warmteoverdrachtscoëfficiënt in $\mathrm{W/(m^2\,K)}$

Dit levert de differentiaalvergelijking:

$$
C \dot{T} = -hA \left( T(t) - T_0 \right)
$$

met $C$ de warmtecapaciteit in $\mathrm{J/K}$.
Herschrijven met:

$$
\tau = \frac{C}{hA}
$$

levert:

$$
-\tau \dot{T} = T(t) - T_0
$$

met als oplossing:

$$
T(t) - T_0 = \left( T(0) - T_0 \right) e^{-t/\tau}
$$

We kunnen hieruit dus concluderen dat $\tau$ de karakteristieke tijdsduur is waarin
de temperatuur van de buis een factor $e$ verlaagt ten opzichte van de omgevingstemperatuur.


**Oefening**

Controleer bovenstaande door vergelijking (oplossing) te substitueren in de differentiaalvergelijking.

**Oplossing**

Differentieer:

$$
T(t) - T_0 = (T(0) - T_0)e^{-t/\tau}
$$

Dit geeft:

$$
\dot{T} = -\frac{1}{\tau}(T(0)-T_0)e^{-t/\tau}
$$

Invullen in:

$$
-\tau \dot{T} = T(t) - T_0
$$

Dit laat zien dat linker- en rechterlid gelijk zijm


We zijn hier voor het gemak uitgegaan van een $h$ die onafhankelijk is van de temperatuur.
We weten echter dat warmtetransport door straling niet lineair gaat, maar als:

$$
\dot{Q}_s = \varepsilon \sigma A \left( T^4 - T_0^4 \right)
$$

Voor kleine temperatuurverschillen ($\Delta T = T - T_0$) is dit te vereenvoudigen tot:

$$
\dot{Q}_s
= \varepsilon \sigma A \left( (T_0 + \Delta T)^4 - T_0^4 \right)
\approx 4 \varepsilon \sigma A T_0^3 \Delta T
$$

Zolang $\Delta T$ dus relatief klein is ten opzichte van $T_0$, kunnen we $h$
dus inderdaad als een constante beschouwen.


**Oefening**

Laat zien dat bovenstaande geldt door eerst $(T_0+\Delta T)^4$ uit te schrijven.

**Oplossing**

$$
(T_0+\Delta T)^4 = T_0^4 + 4T_0^3\Delta T + 6T_0^2\Delta T^2 + \dots
$$

Omdat $\Delta T << T_0$ zijn termen met $\Delta T^2$ en hoger verwaarloosbaar,
daarom blijft alleen de lineaire term ovet


**Oefening**

Hoe groot is de fout in het warmtetransport door straling die we maken?

**Oplossing**

De fout wordt bepaald door de grootte van de term $6T_0^2\Delta T^2$ ten opzichte
van $4T_0^3\Delta T$ en is van orde $\Delta T / T_0$.
Voor het temperatuurbereik in deze proef is deze fout dus klein

**Oefening**

Wat is het buitenoppervlak van de buis?  
Wat is de warmtecapaciteit van de buis?  
Wat is de omgevingstemperatuur?

**Gegeven**

- lengte $h = 152$ mm
- buitendiameter $d = 50$ mm
- binnendiameter $d_\mathrm{in} = 47$ mm
- materiaal: messing
- omgevingstemperatuur: kamertemperatuur

**Antwoorden**

- Buitenoppervlak (cilinder, zonder doppen):

$$
A = \pi d h = \pi \times 0.050 \times 0.152 \approx 2.39 \times 10^{-2}\,\mathrm{m^2}
$$

- Warmtecapaciteit:

$$
C = \rho V c
$$

met:

$$
V = \pi h \left( r_\text{buit}^2 - r_\text{bin}^2 \right)
$$

- Omgevingstemperatuur:

$$
T_0 \approx 20^\circ\mathrm{C}
$$


**Oefening**

Maakt het uit in welke richting je de buis legt? En waarom zit er een dop bij?

**Antwoord**

Ja, de richting beïnvloedt de natuurlijke stroming van de lucht.
De dop voorkomt luchtstroming door de buis en vermindert convectie en daarmee warmtetransport


```python
# Hier de data en de analyse

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

def exp_func(t, A, tau, T_omg):
    # A is verschiltemperatuur met omgeving aan start
    # tau is de karakteristieke tijd voor de koeling
    # T_omg is de omgevingstemperatuur
    return A * np.exp(-t/tau) + T_omg

# Geometrie (eerder bepaald)
buitenoppervlak = 2.39e-2      # m^2

# Warmtecapaciteit (moet overeenkomen met materiaalkeuze)
warmtecapaciteit = 115.2       # J/K  ← bewust open gelaten

dt = 5  # tijdstap in seconde

# Verticaal zonder dop
T_vert_zonder = np.array([
59.5,59.3,59.2,58.8,58.6,58.3,57.8,57.5,57,56.5,56.1,55.8,
55.5,55.1,54.7,54.4,54,53.6,53.2,52.8,52.6,52.3,51.9,
51.5,51.1,50.8,50.5,50.1,49.7,49.4,49,48.6,48.3,48,
47.7,47.4,47,46.7,46.5,46.3,45.9,45.7,45.4,45.1,44.8
])

# Verticaal met dop
T_vert_met = np.array([
58.6,58.3,57.9,57.4,57.2,57,56.3,55.7,55.2,54.6,54.2,53.8,
53.4,53.1,52.7,52.3,52,51.7,51.2,51,50.6,50.2,50,49.7,
49.3,49.1,48.8,48.5,48.2,47.6,47.3,47.1,47,46.7,46.4,
46.2,46
])

# Horizontaal zonder dop
T_hori_zonder = np.array([
55.3,55.1,55.2,55,54.7,54.6,54.2,53.2,52.6,52.1,51.6,51.4,
51,50.9,50.8,50.5,50.1,49.7,49.6,49.3,49.1,48.9,48.6,
48.5,48.4,48.1,47.8,47.6,47.4,47.2,47,46.8,46.6,46.3,
46.2,46,45.6,45.5,45.3,45.1,44.9,44.6,44.5,44.2,44
])

# Horizontaal met dop
T_hori_met = np.array([
60.7,60.3,59.9,59.6,59.2,58.9,58,57.6,57.2,56.7,56.2,56,
55.8,55.1,54.8,53.9,53.2,52.8,52.3,51.8,52,51.5,51,
50.2,49.9,49.6,49.2,49,49.2,49,48.7,48.4,48,47.9,
47.5,47.1,47,46.7,46.6,46.3,46,45.7,45.4,45.1,44.9
])

t_vert_zonder = np.arange(len(T_vert_zonder)) * dt
t_vert_met    = np.arange(len(T_vert_met)) * dt
t_hori_zonder = np.arange(len(T_hori_zonder)) * dt
t_hori_met    = np.arange(len(T_hori_met)) * dt

def fit_en_plot(t, T, titel):
    # beginwaarden: verschil begin- en eindtemperatuur, tau 1000 s, T_omgeving 20°C
    p0 = [T[0] - T[-1], 1000, 20]

    popt, _ = curve_fit(exp_func, t, T, p0=p0, maxfev=10000)
    A, tau, T_omg = popt

    t_fit = np.linspace(0, t[-1], 500)
    T_fit = exp_func(t_fit, *popt)

    plt.figure()
    plt.plot(t, T, 'o', label='meting')
    plt.plot(t_fit, T_fit, '-', label=f'fit, τ = {tau:.0f} s')
    plt.xlabel('tijd [s]')
    plt.ylabel('temperatuur [°C]')
    plt.title(titel)
    plt.legend()
    plt.grid()
    plt.show()

    return tau

tau_vert_zonder = fit_en_plot(t_vert_zonder, T_vert_zonder, "Verticaal zonder dop")
tau_vert_met    = fit_en_plot(t_vert_met, T_vert_met, "Verticaal met dop")
tau_hori_zonder = fit_en_plot(t_hori_zonder, T_hori_zonder, "Horizontaal zonder dop")
tau_hori_met    = fit_en_plot(t_hori_met, T_hori_met, "Horizontaal met dop")

print(tau_vert_zonder, tau_vert_met, tau_hori_zonder, tau_hori_met)

h_vert_zonder = warmtecapaciteit / (tau_vert_zonder * buitenoppervlak)
h_vert_met    = warmtecapaciteit / (tau_vert_met * buitenoppervlak)
h_hori_zonder = warmtecapaciteit / (tau_hori_zonder * buitenoppervlak)
h_hori_met    = warmtecapaciteit / (tau_hori_met * buitenoppervlak)

print("h verticaal zonder dop :", h_vert_zonder, "W/m^2K")
print("h verticaal met dop    :", h_vert_met, "W/m^2K")
print("h horizontaal zonder dop:", h_hori_zonder, "W/m^2K")
print("h horizontaal met dop   :", h_hori_met, "W/m^2K")


```

# Discussie en Conclusie

Uit onze experimenten zien we dat de buis **anders afkoelt afhankelijk van de opstelling**:

- **Verticaal zonder dop:** koelt langzaam af -> klein h.
  Dit komt omdat warme lucht langs de buis opstijgt en er weinig luchtbeweging in de buis is

- **Verticaal met dop:** koelt iets sneller -. groter h.
  De dop zorgt dat lucht binnenin de buis minder beweegt, waardoor warmte meer via het oppervlak naar buiten gaat

- **Horizontaal zonder dop:** koelt sneller dan verticaal zonder dop -> groter h
  Horizontale positie helpt de lucht beter te bewegen langs de buis, waardoor de buis sneller warmte afgeeft

- **Horizontaal met dop:** koelt het snelst -> grootste h.
  Hier helpen zowel de horizontale positie als de dop dat warmte snel wordt afgevoerd via het oppervlak

**Over warmteoverdracht:**

- Het grootste deel van h komt door **convectie**, dus lucht die langs de buis beweeg
- **Straling** levert ook warmteverlies, maar dit is klein bij onze temperaturen
- **Geleiding** zit vooral in de buis zelf en zie je terug in de berekende warmtecapaciteit

**Conclusie:**

1. De tijd τ waarin de buis afkoelt verschilt per opstelling, dat bepaalt hoe groot h is
2. Een dop en horizontale positie zorgen dat de buis sneller afkoelt
3. Newtons afkoelingswet werkt goed voor deze proef, want de buis koelt ongeveer volgens een exponentiële kromme  
4. Als je een buis sneller wilt laten afkoelen, kun je hem horizontaal zetten en zorgen dat lucht goed kan stromen langs de buitenkant
