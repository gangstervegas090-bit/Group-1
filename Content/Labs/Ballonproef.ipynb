# Meten bij constante druk


# Introductie

Volgens [de ideale gaswet](https://nl.wikipedia.org/wiki/Algemene_gaswet) wordt het volume $V$ van een (ideaal) gas gegeven door: 

$$
V = \frac{n R T}{P}
$$

waarin: 

- $n$ het aantal mol gas,
- $R$ de ideale gasconstante,
- $T$ de absolute temperatuur,
- $P$ de druk.

In dit practicum veranderen we de temperatuur en meten we de verandering van het volume van het gas. De proef is met name kwalitatief van aard en laat zien hoe lastig het is een extensieve grootheid als volume te meten.  


# Methode en materialen

## Ontwerp

Om het volume van een hoeveelheid gas bij constante druk te meten is niet zo eenvoudig. Je kunt het gas vrij eenvoudig in een ballon stoppen en dan schatten hoe de diameter van de ballon verandert als functie van de temperatuur, maar dat geeft een relatief grote fout. We maken daarom gebruik van de wet van [Archimedes](https://nl.wikipedia.org/wiki/Wet_van_Archimedes). 

## Materialen

- eenvoudige feestballon
- bekerglas
- tweede bekerglas om mee bij te vullen
- thermometer
- verhittingsplaat
- deksel met vulcilinder met maatstrepen (met iets kleinere diameter dan interne diameter maatbeker)
- (per 5 groepjes) een $10 \mathrm{ml}$ maatcilinder


## Procedure

1. Blaas de ballon op, maar niet verder dan $5 \mathrm{cm}$ in diameter. Deze moet makkelijk in de maatbeker passen. 
2. Knoop de ballon goed dicht zodat er geen lucht kan ontsnappen. 
3. Dompel de ballon onder in de maatbeker met water met behulp van het deksel.
4. Pas het waterniveau aan zodat de meniscus bij een van de onderste maatstrepen van de vulcilinder van het deksel zit. 
5. Let op dat er luchtbellen kunnen plakken aan de ballon wat leidt tot een systematische fout. Verifieer dat je zo min mogelijk systematische fouten maakt en meet op welke maatstreep de meniscus zich bevindt.
6. Verhoog stapsgewijs de temperatuur van het water (en dus de ballon). Let op dat je de temperatuur maximaal een graad of 20 kan verhogen.
7. Laat het geheel na elke temperatuurtoename een minuut 'rusten' om thermisch evenwicht te bereiken.
8. Meet na elke temperatuurtoename de temperatuur en de positie van de meniscus.

Het verschiloppervlak tussen de binnendiameter van de maatbeker en de buitendiameter van de vulcilinder is $10.0 \mathrm{cm}^{2}$. Je kunt die handmatig kalibreren met behulp van de kleine maatcilinder die in het lokaal aanwezig is. De maatstreepjes die op de vulcilinder zijn gekerfd zitten op een onderlinge afstand van $1.0 \mathrm{mm}$.


### Opdracht 1: Analyseer de data
Plot de volumeverandering van de ballon als functie van de temperatuur. Zie je hier een lineair verband? Bepaal het startvolume van de ballon door te extrapoleren naar een absolute temperatuur van $0 \mathrm{K}$.

### Opdracht 2: Verbeterde meting
- Maak de maatbeker leeg en vul deze opnieuw met water tot een van de onderste maatstrepen van de vulcilinder (dus zonder ballon). 
- Verhit de maatbeker weer stapsgewijs en meet na elke stap de temperatuur en de hoogte van de meniscus.

Kun je nu met behulp van deze tweede meting een betere afschatting geven voor het startvolume van de ballon?


# Resultaten


```python
import numpy as np
import matplotlib.pyplot as plt

# Temperatuur [°C] en meethoogte [mm]
temps = np.array([16.7, 26.9, 32.3, 36, 47.5, 58.1, 81, 96.3])
hoogtes = np.array([325, 340, 350, 355, 395, 405, 450, 500])  # in ml equivalent
```

```python
# Afmetingen in cm
diam_beker = 7.66  # cm
diam_deksel = 6.77 # cm
opp_water = 1/4 * np.pi * (diam_beker**2 - diam_deksel**2)  # cm^2

# Volumes van ballon berekenen t.o.v. starthoogte
volumes = (hoogtes - hoogtes[0]) * opp_water  # cm^3

# Absolute temperatuur in K
temps_abs = temps + 273
```

```python
# Lineaire fit
a = np.polyfit(temps_abs, volumes, 1)
func = np.poly1d(a)

# Plot
plt.figure(figsize=(7,5))
plt.plot(temps_abs, volumes, 'ob', label='Data')
plt.plot(temps_abs, func(temps_abs), '--r', label='Lineaire fit')
plt.xlabel('Temperatuur [K]')
plt.ylabel('Volume [cm³]')
plt.title('Volumeverandering van ballon als functie van temperatuur')
plt.legend()
plt.grid(True)
plt.show()

print(f"Lineaire fit: V = {a[0]:.3f} T + {a[1]:.3f}")
print(f"Startvolume (doortrekken naar 0 K): {func(0):.2f} cm³")
```

# Discussie en conclusie

**Discussie:**

Uit de grafiek blijkt dat het volume van de ballon toeneemt als de temperatuur stijgt. Het verband lijkt ongeveer lineair te zijn, wat overeenkomt met de verwachting van de ideale gaswet. Kleine afwijkingen kunnen komen door luchtbellen aan de ballon of het uitzetten van het water in de maatbeker.

**Conclusie:**

Het experiment laat zien dat het volume van de ballon bij constante druk lineair toeneemt met de absolute temperatuur, zoals voorspelt door de ideale gaswet. Door dit verband door te treken kon het startvolume bij 0 K worden geschat, hoewel deze waarde theoretisch is en beinvloed kan worden door kleine meetfouten zoals luchtbellen en het uitzetten van water
